```python
drone = {
    "brand": "DJI",
    "model": "Mavic 3",
    "max_speed_kph": 75,
    "flight_modes": ["Sport", "Normal", "Cine", "Waypoint"],
    "camera": {
        "resolution": "5.1K",
        "fps": 60,
        "sensor": "CMOS",
    },
    "range_km": 15,
    "weight_grams": 895
}

print(drone)
```

# copy Method

```python
a = drone.copy()

print(a)
```

# clear method

```python
a.clear()
print(a)
```

# get method

```python
d = drone.get("model")

print(d)
```

# setdefault method

```python
x = drone.setdefault("price",50000)

print(x)
```

# keys

```python
z = drone.keys()

print(z)
```
