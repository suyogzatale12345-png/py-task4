```python
coffee_machine = {
    "brand": "De'Longhi",
    "type": "Espresso",
    "water_capacity_liters": 2,
    "coffee_types": ["Latte", "Cappuccino", "Americano", "Flat White"],
    "settings": {
        "grind_level": 7,
        "temp": "High",
        "milk_froth": "Auto"
    },
    "pressure_bar": 15,
    "auto_clean": "True"
}

print(coffee_machine)
```

# values method

```python
z = coffee_machine.values()

coffee_machine["pressure_bar"] = 20
print(z)
```

# fromkeys method

```python
b = ("tabala","banjo", "dj")
c = 0
d = dict.fromkeys(b)

print(d)
```

# item method

```python
l1 = coffee_machine.items()

coffee_machine["grind_level"] = 10

print(l1)
```

# copy method

```python
f1 = coffee_machine.copy()
print(f1)
```

# clear method

```python
f1.clear()
print(f1)
```
