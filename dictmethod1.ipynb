# Dictionary  Method

```python
car_info = {
    "brand": "Ford",
    "model": "Mustang",
    "Year": 1964,
    "color": "Red",
    "engine": "v8"
}

print(f"car_info = {car_info}, lenth of car_info = {len(car_info)}, types of car_info = {type(car_info)}")
```

# clear() Method

```python
car_info.clear()
print(car_info)
```

# copy() Method

```python
car_data = car_info.copy()

print(f"after car_info copy = {car_data}")
```

# get method


```python
x = car_info.get("model")

print(x)


print()
print()

print(car_info)
```

# iteam

```python
x = car_info.items()

car_info["color"] = "Red"

print(x)
```

# keys

```python
x = car_info.keys()

car_info["price"] = 100000


print(x)
```
