1. Library Management
book_name: "Python Basics"
total_copies: 50
shelf_number: 12
pages: 320
author_name: "John Smith"
published_year: 2023
loan_duration_days: 14

```python
lib = {
    "book_name": "python_basic",
    "total_copies": 50,
    "shelf_number": 12,
    "pages": 320,
    "auther": "john smith",
    "published_year": 2023,
    "loan_duration_days": 14  
}

print(lib)
```

```python
print(f"lib ={lib} , len of library = {len(lib)}, types of = {type(lib)}")
```

# clear 

```python
lib.clear()

print(library)
```

# copy method

```python
library = lib.copy()

print(library)
```

# get Method

```python
a = library.get("book_name")

print(a)
```

# item Method

```python
b = library.items()

print(b)
```
