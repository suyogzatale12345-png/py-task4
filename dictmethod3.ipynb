```python
user = {
    "id": 101,
    "name": "Harsh",
    "role": "Dev",
    "active": "True",
    "pings": 42
}

print(f"user = {user}, len of user = {len(user)}, types of user = {type(user)}")
```

# values

```python
c = user.values()

user["pings"] = 20

print(c)
```

# update

```python
user.update({"age": 28})

print(user)
```

# get 

```python
u = user.get("role")

print(u)
```
